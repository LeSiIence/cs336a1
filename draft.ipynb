## About einops

```python
import torch
from einops import rearrange, einsum
batch = 2
sequence = 4
d_in = 3
d_out = 5

D = torch.randn(batch, sequence, d_in)
A = torch.randn(d_out, d_in)
Y = D @ A.T

Y = einsum(D, A, "batch seqence d_in, d_out d_in -> batch seqence d_out")
print(Y)
```
